# Mercados y comercialización de la energía eléctrica

## La bolsa de energía

**Cadena de la energía eléctrica**
1. Generación
2. Transmisión
3. Distribución
4. Consumidor

- La energía fluye de la generación al consumidor, mientras que el dinero fluye en la dirección contraria.
    - El consumidor paga por el servicio.
    - El dinero debe surtir para las demás instancias.
    - Se le paga al comercializador.
    - El comercializador distribuye el dinero, incluyendo a la regulación, vigilancia, operación, etc.
    - Los generadores operan en el Mercado mayorista. El comercializador determina a qué agente le compra la energía. Esto está dentro del mercado mayorista

- Mercado Mayorista
    1. Generación
    2. Agentes
    3. Recursos

**Mercados mayoristas de energía**
- Agentes: Dueño / Agente generador
- Plantas (de generación)
- Unidades: distintas máquinas alrededor de la misma planta (central)

**Oferta de generación: Actualizacón cada 24 hrs**
- Cada agente oferta, para cada unidad de cada una de sus plantas.
    - D: Disponibilidad de energía. Es la cantidad de energía que puede suministrar para cada hora del día, del día siguiente.
    - P: Precio al que vente la energía. Es un mismo precio para todo el día, para cada planta. El precio permanece secreto por 30 días.
- También puede ofertar otros servicios. P.ej. Control de generación

**El mecanismo de la bolsa. Para cada una de las 24 horas del día siguiente:**

- Agente 1:
    - Planta 1-1: P, D
    ...
    - Planta 1-p: P, D

- Para cada una de las horas:
    - Si formamos rectángulos, donde P es la altura y D la base.
    - Se pueden representar en una gráfica.
    - Las ofertas se ordenan por precio, de la más barata a la más cara.
    - Se pinta la Predicción de la demanda dentro de la gráfica.
    - Las plantas que se encuentren en el lado izquierdo de la franja de la demanda, son las que despachan energía.
    - El precio de la última oferta determina el precio de la bolsa para esa hora.
    - Todos los agentes son remunerados al mismo precio de bolsa.
    - Un pequeño cambio en la demanda...
    - Generan cambios importantes en el precio de la energía.
    - Si la disponibilidad de la planta cambia, también puede cambiar el precio de la energía.

## El precio de la energía en el mercado mayorista

**El mecanismo de la bolsa. Para cada una de las 24 horas del día siguiente:**

- El precio de la última oferta  determina el precio de bolsa para esa hora.

**Predespacho ideal**

- Esto da como resultado el predespacho ideal.
    - Cantidad de energía que la planta deberá general al día siguiente en cada una de las franjas horarias.

**Los ajustes**
- Despacho ideal
- Redespacho: Restricciones operativas del sistema de potencia. Es posible que existan situaciones como congestión en las líneas que impida que se realice, efectivamente, esa asignación de generación para atender la demanda nacional.
- Operación en tiempo real. Cambios abruptos en la demanda, salida del servicio de una planta o línea de transmisión.
- Reconciliaciones:  Ajustes de dinero previstos, es decir, lo que se había previsto en el despacho ideal y lo que sucedió en la realidad.

**Precios de bolsa**
- Precio máximo del día.
- Precio promedio del día.
- Precio ponderado del día.

**Otros precios**
- Precio de volsa
- Precios de contratos bilaterales:
    - Mercado regulado: Protegen a los usuarios pequeños que tienen bajos consumos.
    - Mercado no regulado:  el de los clientes que necesitan gran cantidad de energía, como industrias. Se les permite negociar con unas reglas especiales con los comercializadores.

**Volatibilidad de los precios**
- Resolución por hora.
- Resolución por día en un mes. 
- Resolución mensual en el año.

- Se puede graficar el precio promedio ponderado de la bolsa nacional
- Se puede graficar la volativilidad diaria móvil del precio de bolsa, es decir el porcentaje de incremento o decremento con respecto al día anterior.

- En varios años, resolución diaria.

## Tarifas. El precio de la energía para el usuario

**Precio de la energía al usuario**

$$
Costo = Costo unitario \times Consumo
$$
$$
C = CU \times Q
$$

**Caso Colombiano**
$$
CU = G + T + D + Cv + Pr + R
$$
$$
Costo Unitario = Generación + Transmisión + Distribución + Comercialización + Pérdidas + Restricciones
$$

- Generación. Promedio ponderado de precios de compras en
    - La bolsa de energía
    - Contratos bilaterales
    - Subastas de Energía
- Transmisión. Cargo por el uso del sistema de transmisión nacional.
- Distribución. Cargo por el uso de:
    - El sistema de transmisión regional.
    - El sistema de distribución local.
- Comercialización. Cargo por conceptos de:
    - El margen de utilidad al comercializador
    - La contribución a las entidades de control y vigilancia (CREG y SSPD)
    - Los costos de servicios de operadores técnicos y financieros (CND y ASIC)
    - Los costos de garantías
- Pérdidas. Cargo por:
    - Pérdidas de energía en el sistema de transmisión.
    - Pérdidas de energía en el sistema de distribución.
    - Programas de inversión para reducción de pérdidas no técnicas.
- Restricciones. Cargo por operación subóptima del sistema, debida a restricciopnes técnicas operativas, tales como:
    - Congestión de líneas o subestaciones
    - Indisponibilidad de líneas o subestaciones
    - Consideraciones de seguridad y robustez

El costo de la energía cambia de un día a otro y de una región a otra.

**Peso de los componentes tarifarios. Promedio nacional**

Los precios de la Generación y la Distribución son los que tienen mayor aporte al costo unitario de la energía. También son las que tienen mayores variaciones.

Los componetes de Transmisión son muy constantes.

**Curiosidades del caso colombiano. Estratificación socioeconómica**

- Clasificación en 6 estratos de los inmuebles residenciales que deben recibir servicios públicos.
- Permite cobrar de manera diferencial los servicios públicos domiciliarios permitiendo asignar subsidios y cobrar contribuciones en esta área.
- La mayor parte de los subscriptores se encuentran en el estrato 1 y 2. Donde el 1 es de nivel socioeconómico bajo y el 6 de nivel socioeconómico alto.
- El estrato 6 muestra la menor cantidad de subscriptores.
- El estrato 1 al 3, recibe subsidios.
- El estrato 6 y 6, realizan contribuciones de excedentes.
- El déficit se cubre por parte del gobierno colombiano, que viene de los impuestos.
