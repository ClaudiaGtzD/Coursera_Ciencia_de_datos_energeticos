# Series de tiempo

## Series de tiempo

Una de las columnas de los datos, representa el tiempo en el que fueron tomados los demás valores.
- Datos ordenados cronológicamente
- Datos igualmente espaciados en el tiempo

**Horizontes de tiempo**

![Horizontes de tiempo](Horizontes_tiempo.PNG)

- Calidad de potencia -> Segundos, minutos
- Eficiencia energética de equipos -> Minutos, horas
- Consumo de energía -> Horas, días
- Despacho económico ->  Horas, días
- Precios de la energía -> Dias, meses
- Planificación de la expansión -> Años, décadas

**Análisis de series de tiempo**
- **Análisis descriptivo**: ¿Cómo es?
- **Análisis explicativo**: ¿Por qué es así?

**Análisis descriptivo de una serie de tiempo. ¿Cómo es?**

$$
Serie = Tendencia + Estacionalidad + Aletoriedad
$$

![Descomposición de una serie de tiempo](Decomposition.PNG)

**Tendencia. Promedio de ventana móvil**

- Promedio de la ventana. Por ejemplo, cada 5 datos.
- La tendencia depende del tamaño de la ventana.
- Suaviza los datos.
- Entre mayor sea el tamaño de la ventana, mayor es la suavización.

**Estacionalidad**
1. $\bar{X}$: Tendencia
2. Identificar ciclos.
3. Comportamiento estacional: $I_k$: promedio de los $x/\bar{x}$ de su ciclo
4. Aletoriedad: $s_t=(I_t-1)x_t$

**Análisis descriptivo de una serie de tiempo. Descomposición multiplicativa**
$$
Serie = Tendencia \times Estacionalidad \times Aletoriedad
$$

![Descomposición multiplicativa](Descomposicion_multiplicativa.PNG)

$$
x=\bar{x} \times S \times R
$$
$$
ln(x) = ln(\bar{x}) + ln(S) + ln(R)
$$

**Análisis descriptivo de una serie de tiempo. Anpalisis de Fourier**

- Dominio del Tiempo -> Transformada de Fourier -> Dominio de la frecuencia
- Reescritura de la serie de tiempo en la suma de señales sinuisoidales.

![Análisis de Fourier](Analisis_Fourier.PNG)

- Con esto se puede usar para determinar ciclos.
- Diseñar filtros que estudien comportamientos de ciclos.


## Identificación de sistemas dinámicos

**Búsqueda de explicaciones**
- ¿Por qué toma esos valores -> ¿Las otras variables pueden ser la causa?
- Lo nuevo: Sistemas con memoria

**Memoria: cuando el pasado condiciona el presente. Ejemplos de fenómenos:**
- Sistemas físicos:
    - Almacenamiento de energía
    - Acumulación de materia
- Sistemas económicos:
    - Acumulación de capital
- Sistemas demográficos:
    - Crecimiento de la población

**Sistemas con memoria - Sistemas dinámicos. Representación del tiempo**

- Sistemas continuos $t\in \mathcal{R}$
    - Ecuaciones diferenciales:
    $$
    f(t), \frac{df(t)}{dt}, \int_{}^{}f(t)dt
    $$
- Sistemas discretos $k\in \mathcal{Z}$
    - Ecuaciones de diferencias
    $$
    f(k), f(k-1), f(k+1)
    $$
- Transformación bilineal
- Tiempo de muestreo

## Métodos de identificación de sistemas dinámicos

**Identificación de sistemas. Dos caminos**
- Modelos de Predicción del Error
$$
    A(q)y(k)=\frac{B(q)}{F(q)}u(k)+\frac{C(q)}{D(q)}e(k)
$$
    - A, B, C, D, F son polinomios que representan retardos de la señal.
- Métodos de Subespacio de Estado
$$
\begin{bmatrix}
X(k+1) \\
Y(k)
\end{bmatrix}
=
\begin{bmatrix}
A & B \\
C & D
\end{bmatrix}
\begin{bmatrix}
X(k) \\
U(k)
\end{bmatrix}
+
\begin{bmatrix}
v(k) \\
w(k)
\end{bmatrix}
$$

    - **A, B, C, D** son matrices.  
    - **X, w, v** son vectores de variables.

**Modelos de predicción del error. Modelo ARX**

Familia de posibles modelos representados por la siguiente ecuación:
$$
A(q)y(k) = \frac{B(q)}{F(q)}u(k) + \frac{C(q)}{D(q)}e(k) \quad \Rightarrow \quad A(q)y(k) = B(q)u(k) + e(k)
$$

$$
A(q)y(k) = B(q)u(k) + e(k)
$$

*A, B* son polinomios que representan retardos de la señal. Por ejemplo:

$$
\begin{aligned}
A(q) &= 1 + a_1 q^{-1} \quad &\Rightarrow \quad A(q)y(k) &= y(k) + a_1 y(k-1) \\
B(q) &= b_1 q^{-1} + b_2 q^{-2} \quad &\Rightarrow \quad B(q)u(k) &= b_1 u(k-1) + b_2 u(k-2)
\end{aligned}
$$

$$
y(k) + a_1 y(k - 1) = b_1 u(k - 1) + b_2 u(k - 2) + e(k)
$$

*e* es el error.

Hay que predecir el error *e* para que forme parte del modelo como una de las variables.

ARX = AutoRegresivo con variables Exógenas

**Métodos de subespacio de  estado. Las variables de estado**

$$
\begin{bmatrix}
X(k+1) \\
Y(k)
\end{bmatrix}
=
\begin{bmatrix}
A & B \\
C & D
\end{bmatrix}
\begin{bmatrix}
X(k) \\
U(k)
\end{bmatrix}
+
\begin{bmatrix}
v(k) \\
w(k)
\end{bmatrix}
$$

- Las *Variables de Estado* $X(k)$ son internas.
- Las variables de estado conforman un espacio vectorial denominado *Espacio de Estado*.
- El número de variables de estado determina el *orden* del sistema.
- Hay infinitos espacios de estado válidos para un mismo sistema.
- A cada espacio le corresponde un juego de matrices *A, B, C, D*.
- En algunos de esos espacios de  estado es más evidente cuáles variables aportan poco a la dinámica del sistema (valores singulares de una matriz)

Algoritmo:
1. A partir de las mediciones de *U* y *Y*, encontrar un buen espacio vectorial.
2. Determinar cuántas variables de estado son suficientes para explicar la dinámica del sistema.
3. Obtener un *subespacio de estado*, descartando las variables de estado innecesarias.
4. Si se requiere, buscar otro espacio vectorial adecuado para la aplicación.

**Una comparación entre los dos caminos**

| **Predicción del error**               	| **Subespacio de estado**                       	|
|----------------------------------------	|------------------------------------------------	|
| Concebido para sistemas SISO           	| Concebido para sistemas MIMO                   	|
| Múltiples esquemas                     	| Un esquema unificado                           	|
| El orden se determina mediante pruebas 	| El orden se determina dentro del procedimiento 	|
| Requiere pocas mediciones              	| Altas exigencias para las mediciones           	|
