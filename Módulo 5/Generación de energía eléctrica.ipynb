#  Generación de energía eléctrica

## Tecnologías para la generación de energía eléctrica

**Transformaciónes de energía**

- Electroquímica: pilas
- Efecto fotoeléctrico: Paneles solares. Energía solar -> Energía eléctrica
- Electromagnetismo: Energía mecánica -> Energía eléctrica
- Hidroeléctrica a filo de agua: Caudal de Agua -> Tubería -> Turbina -> Generador / E. Potencial gravitacional -> E. Hidráulica -> E. Cinética rotacional -> E. Eléctrica
- Hidroeléctrica con embalse: Embalse de Agua -> Tubería -> Turbina -> Generador / E. Potencial gravitacional -> E. Hidráulica -> E. Cinética rotacional -> E. Eléctrica
- Termoeléctrica con carbón:  Horno -> Vapor -> Turbina -> Generador / E. Química -> E. Térmica -> E. Cinética rotacinal -> E. Eléctrica

## Tipos de plantas de generación de energía eléctrica

**Plantas generadoras basadas en inducción electromagnética**
- Termonuclear

## Recursos energéticos

**No todos los recursos energéticos son iguales**

**El panorama mundial - Energía total**
- Se usan muchos recursos fósiles para generar energía (Aceite, nuclear, carbón, gas natural)
- Se usan muchos recursos fósiles para generar energía eléctrica, pero sobresale la energía hidroeléctrica.
- Hay un pequeño rango de energía fotovoltaica y eólica

**Viento - densidad de potencia media ($v / m^2$)**
- El viento se acumula en regiones costeras.
- Perfil vertical del viento
    - A mayor altura, mayor velocidad del viento, por lo que hay mayor potencia

**Sol - irradiación normal directa. promedio anual**
- En el ecuador no hay una alta irradiancia
- Afectaciones como la altitud y la nubosidad afectan al nivel de irradiancia
- Zonas desérticas destacan como buenas opciones

**Gasoductos**
- Infraestructura para el aprovechamiento del gas natural en ciertos países
- México está muy chido para eso

## Recursos energéticos de un país. El caso de Colombia

**Potencial hidroenergético**
- Caudal
- Altura
- Tecnología
- Restricciones ambientales
- Clasificación por cuencas

**Estacionalidad de la radiación solar**
- En el norte hay potencial

**Estacionalidad del viento**
- La variación de los 12 meses beneficia la mayor parte del año.
- Durante del día, la mayor cantidad de viento sucede durante el día.


**Costos de recursos renovables en el rango de los fósiles**
- Con respecto a los costos de fósiles, la tecnología solar es muy cara, ya que a veces ocupas lugares remotos y requieres hacer toda la interconexión.
- La generación eólica en mar es más costosa que en tierra.

**Evolución de los costos de recursos renovables**
- Del 2010 al 2020, hay una gran disminución de costos, lo que lo hace competitivo a los fósiles.
- La generación eólica también ha tenido disminuciones.
- Las energías geotérmicas han tenido crecimiento en costos.
