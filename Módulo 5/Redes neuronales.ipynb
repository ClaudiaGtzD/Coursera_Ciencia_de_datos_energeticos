## Aproximadores de funciones

**Aproximadores universales de funciones**

- $x \longrightarrow  [y=f(x)] \longrightarrow  y = f(x)$
- $x \longrightarrow  [Aproximador] \longrightarrow  \bar{y}=\Phi(x)$
- ¿Puedo diseñar un aproximador tal que $y= \bar{y}$?
- ¿Puedo diseñar un aproximador tal que $y\approx \bar{y}$?
- ¿Puedo diseñar un aproximador tal que $\left|y-\bar{y} \right|<\epsilon$, para cualquier $\epsilon > 0$?

**¿Puedo usar un aproximador que aproxime cualquier función?**
1. ¿Para cualquier función **existe** al menos un miembro de la familia que la aproxime con la precisión que se quiera?
    - Teoremas de existencia
2. Si existe, ¿cómo lo encuentro? ¿cómo lo **diseño**?
    - Métodos de búsqueda

**Para qué aproximar funciones?**
- La función es desconocida pero se puede conocer su comportamiento.
    - Reconocimiento de patrones, identificación de sistemas, observadores de estado, descubrimiento de conocimiento, etc.
- La función es muy complicada de calcular o analizar.


## Estructura del perceptrón multicapa

Perceptrón multicapa
- Capa de entrada, con los nodos con las variables de entrada
- Capa oculta: número de nodos, función de activación, parámetros entre capas
- Capa de salida con la variable aproximada de salida

## Entrenamiento del perceptrón multicapa

**¿Cómo determinar los parámetros?**
- Tenemos una colección de datos disponibles.
- Seleccionamos una función de activación y un número de nodos.
- Asignamos de manera arbitraria, el valor de los demás parámetros, el peso, los bias, que se irán moviendo.
- Se determina el error total
- Se ajustan los parámetros, basado en el error (backpropagation).
    - Técnica del gradiente descendiente.
