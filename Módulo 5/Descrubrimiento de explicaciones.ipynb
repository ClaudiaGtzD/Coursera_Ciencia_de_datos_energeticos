## Árboles de decisión

**El problema a resolver: Búsqueda de explicaciones**

¿Por qué toma esos valores? -> ¿Las otras variables pueden ser la causa?

Aprendizaje supervisado

**Dos versiones del problema:**
- ¿Qué es lo que se quiere explicar?
    - *Clasificador*: y es ordinal. Ej: árboles de decisión
    - *Regresor*: y es numérico. Ej: regresión lineal

**Árboles de decisión**
- Variables *x*: Temperatura, Humedad
- Variable *y*: Confort (Frío, Confortable, Insoportable, etc.)

## Regresores lineales

**Regresiones**

$X=\left\{ x_1,x_2,\cdots x_m \right\}$

$X, y$ son valores numéricos

Encontrar $f$ a partir de los datos

- x -> y
- Regresores -> Respuesta
- Variables explicativas -> Variable explicada
-> Entradas -> Salida

**Una aproximación**
- Escoger una familia de funciones $\mathcal{F}$ parametrizada:
$$
\mathcal{F}:\left\{ f(X;\theta ) \right\}, \theta =\left\{ \theta _1, \theta _2, \cdots \theta _p \right\}
$$
- Escoger una medida del error cometido al estimar $y$ con una $f(X;\theta )$, p.ej.
$$
\varepsilon(\theta )=\frac{1}{n}\sum_{0}^{n}(y_j-f(X_j;\theta ))^2
$$
- Encontrar los parámetros \bar{\theta } que minimizan el error
$$
\bar{\theta }=arg_{min}\left\{ \varepsilon(\theta ) \right\}
$$

**Hallar $y=f(x)$. Una entrada y una salida**
- $\mathcal{F}$: conjunto de todas las rectas
$$
\mathcal{F}:\left\{ f(x)=mx+b \right\}, \theta=\left\{ m,b \right\}
$$
- Usar el error medio cuadrático
$$
\varepsilon(\theta )=\frac{1}{n}\sum_{0}^{n}(y_j-(mx_j+b ))^2
$$
- Los parámetros que minimizan el error satisfacen
$$
\begin{cases}
m \sum_{j=1}^{n} x_j + b n = \sum_{j=1}^{n} y_j \\\\
m \sum_{j=1}^{n} x_j^2 + b \sum_{j=1}^{n} x_j = \sum_{j=1}^{n} x_j y_j
\end{cases}
$$
- $m, b$ son los parámetros de la mejor solución
