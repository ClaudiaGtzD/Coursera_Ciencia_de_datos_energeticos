# Transporte de energía eléctrica

## Los cambios de tensión

**La cadena de electricidad**
- Generación: Transformación de una fuente de energía en energía eléctrica
- Transmisión: grandes distancias con grandes cantidades de energía
- Distribución: Reparto de energía
- Consumidor: Gastar esa energía

**Transmisión**
- Fuente de energía: Lugar donde se general la energía eléctrica
- Medio de transporte: red de transmisión
- Uso de energía: subestaciones

**Distribución**
- Fuente de energía: subsestaciones donde la transmisión depositó la energía
- Medio de transporte: red de distribución
- Centros de carga (hogares, comercios, etc.)

**Preocupaciones a atender en el transporte de energía**
- Capacidad de transporte
- Calidad de potencia
- Eficiencia energética
- Seguridad:
    - Sobrecargas
    - Cortoscircuitos
    - Sobretensiones
- Atención a fallas

**Capacidad de transporte**
- Fuente = $V$
- Medio = Resistencias $\frac{R_C}{2}\pm\frac{V_C}{2}$
- Carga = $R_L$

$I_L=\frac{V}{R_T}=\frac{V}{R_C+R_L}$

$V_C=I_L R_C=V\frac{R_C}{R_C+R_L}$

$P_C=I_L V_C=V^2\frac{R_C}{(R_C+R_L)^2}$

¿Cómo disminuir $I_L$, $P_C$?
- Agregar Transformador Elevador entre la Fuente y el Medio
- Agregar Transformador Reductor entre el Medio y la Carga

**Transformador eléctrico**
- Funciona con electromagnetismo
- A menor arrollamiento: menor tensión, mayor corriente
- A mayor arrollamiento: mayor tensión, menor corriente
- Potencia = Tensión * corriente, es decir conservación de potencia
- En un caso donde hay arrollamiento menor en la izquierda y arrollamiento mayor a la derecha, hay un incremento de tensión, al costo de disminuir la corriente.
- ¿Qué es preferible?
    - Tensión baja y corriente alta ó
    - Tensión alta y corriente baja
    - Depende:
        - En el consumo, es mejor trabajar con tensiones bajas (en el orden de los 100V)
        - En el transporte de energía, es mejor trabajar con corrientes bajas, para permitir conductores más delgados que soporten

**Capacidad de Transporte, niveles de tensión**
- Generación: Tensión $<10kV$
- Transmisión: Tensión $>200kV$, en distancia $>100km$.
- Distribución: Tensión $<20kV$, en distancia $\thickapprox10km$
- Consumo: Tensión $<0.5kV$

## El calentamiento y capacidad de transporte

**Capacidad de Transporte**
Calentamiento:
- Sol
- Efecto Joule = $I^2R$, a mayor corriente, mayor calentamiento

Enfriamiento
- Viento
- Radiación

Al haber calentamiento, la línea de transporte se dilata. Es importante estudiar la caida de línea (flecha).

**El conductor**
- Seleccionar el conductor de una línea
- Material interno de acero
- Material externo con buena propiedad eléctrica
- A mayor grosor, mayor capacidad de transporte

**Criterio económico de diseño**
- Costos de instalación: a mayor diámetro, mayor costo
- Costos de operación: a mayor diámetro, menor el costo. (las pérdidas por el efecto Joule disminuye)
- Se busca el diámetro óptimo que minimice lo anterior.

**Capacidad de transporte. Diseño**
- Estimar el calor máximo soportado por la línea.
- Tomar el peor escenario, cuando el calor proporcionado por el sol es el máximo y el enfriamiento por el viento es mínimo. Margen: &Q_max-Q_s+Q_v& Cargabilidad fija
- Calcular el margen: &Q_max-Q_s+Q_v& para cada hora para determinar el margen durante todo el día. Gana el margen mínimo. Cargabilidad dinámica
- La diferencia de Cargabilidad fija y la dinámica es la ganancia de utilizar la dinámica en vez de la fija
- Para cambiar a cargabilidad dinámica, depende de la capacidad de predecir condiciones ambientales y reaccionar ante cambios repentinos.

## Transmisión y distribución de energía eléctrica

**Diferencias entre Transmisión y Distribución**
- Distancia
- Potencia
- Tensión
- Topología
- Control
- Medición
- Protección

**Niveles de tensión, distancias y potencias**

|              | Tensión  | Distancia      | Potencia      |
|--------------|----------|----------------|---------------|
| Transmisión  | $>200kV$ | $>100km$       | $>10MW$       |
| Distribución | $>20kV$  | $\backsim10km$ | $\backsim1MW$ |

**Subestaciones**
Punto de acople entre el sistema de transmisión a un sistema de distribución.

Transmisión: 1 sistema trifásico

Distribución: múltiples sistemas trifásicos

Se usa un transformador para disminuir la tensión recibida de la red de transmisión.

**Sistema de distribución: circuitos primarios y secundarios**
Primarios (Media tensión) Línea de distribución (1 sistema trifásico) -> transformador (reduce tensión) -> múltiples usuarios (sistemas de 4 hilos) Secundarios (baja tensión)

**Transmisión vs Distribución: Topología**
- Transmisión: Enmallada, es decir todo está interconectado entre todas las subestaciones
- Distribución: Radial: Solo llega a las subestaciones donde se consume, no se suelen interconectar

**Transmisión vs Distribución: Control**
- Transmisión:
    - Generación:
        - Potencia Activa - Frecuencia
        - Potencia Reactiva - Tensión
    - Transformación:
        - Sin carga vs con carga
    - Compensación reactiva:
        - Sin carga vs con carga
        - Dinámica
    - Conexión y desconexión:
        - Protección
        - Deslastre de carga
- Distribución:
    - Transformación:
        - Solo en las subestaciones
    - Compensación reactiva:
        - Líneas largas
    - Conexión y desconexión:
        - Protección

**Transmisión vs Distribución: Mediciones**
- Transmisión:
    - En subestaciones:
        - Magnitudes de Tensiones
        - Magnitudes de Corrientes
        - Potencia activa, reactiva y aparente
        - Energía
        - Frecuencia
        - Unidades de medición fasorial - PMU
            - Fasores de Tensiones
            - Fasores de Corrientes
            - Frecuencia
- Distribución:
    - En subestaciones:
        - Magnitudes de Tensiones
        - Magnitudes de Corrientes
        - Frecuencia
        - Potencia activa, reactiva y aparente
        - Energía
    - En los consumidores:
        - Consumo mensual de energía
        - Instrumentos de medición avanzada - AMI
            - Magnitudes de Tensiones
            - Magnitudes de Corrientes
            - Potencia activa, reactiva y aparente
            - Energía

**Nuevos (y viejos) retos para los sitemas de distribución**
- Autogeneración y flujo bidireccional de energía
- Cargabilidad dinámica y ajuste en línea de las protecciones
- Crecimiento no planificado de la demanda. Impacto de los vehículos eléctricos
- Localización de fallas en sistemas subterráneos
- Predicción de fallas y gestión de mantenimiento
