# Transporte de energía eléctrica

## Los cambios de tensión

**La cadena de electricidad**
- Generación: Transformación de una fuente de energía en energía eléctrica
- Transmisión: grandes distancias con grandes cantidades de energía
- Distribución: Reparto de energía
- Consumidor: Gastar esa energía

**Transmisión**
- Fuente de energía: Lugar donde se general la energía eléctrica
- Medio de transporte: red de transmisión
- Uso de energía: subestaciones

**Distribución**
- Fuente de energía: subsestaciones donde la transmisión depositó la energía
- Medio de transporte: red de distribución
- Centros de carga (hogares, comercios, etc.)

**Preocupaciones a atender en el transporte de energía**
- Capacidad de transporte
- Calidad de potencia
- Eficiencia energética
- Seguridad:
    - Sobrecargas
    - Cortoscircuitos
    - Sobretensiones
- Atención a fallas

**Capacidad de transporte**
- Fuente = $V$
- Medio = Resistencias $\frac{R_C}{2}\pm\frac{V_C}{2}$
- Carga = $R_L$

$I_L=\frac{V}{R_T}=\frac{V}{R_C+R_L}$

$V_C=I_L R_C=V\frac{R_C}{R_C+R_L}$

$P_C=I_L V_C=V^2\frac{R_C}{(R_C+R_L)^2}$

¿Cómo disminuir $I_L$, $P_C$?
- Agregar Transformador Elevador entre la Fuente y el Medio
- Agregar Transformador Reductor entre el Medio y la Carga

**Transformador eléctrico**
- Funciona con electromagnetismo
- A menor arrollamiento: menor tensión, mayor corriente
- A mayor arrollamiento: mayor tensión, menor corriente
- Potencia = Tensión * corriente, es decir conservación de potencia
- En un caso donde hay arrollamiento menor en la izquierda y arrollamiento mayor a la derecha, hay un incremento de tensión, al costo de disminuir la corriente.
- ¿Qué es preferible?
    - Tensión baja y corriente alta ó
    - Tensión alta y corriente baja
    - Depende:
        - En el consumo, es mejor trabajar con tensiones bajas (en el orden de los 100V)
        - En el transporte de energía, es mejor trabajar con corrientes bajas, para permitir conductores más delgados que soporten

**Capacidad de Transporte, niveles de tensión**
- Generación: Tensión $<10kV$
- Transmisión: Tensión $>200kV$, en distancia $>100km$.
- Distribución: Tensión $<20kV$, en distancia $\thickapprox10km$
- Consumo: Tensión $<0.5kV$
