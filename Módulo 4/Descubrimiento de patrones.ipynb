**Algoritmos de arupamiento**
- Matriz de distancias: indica la distancia entre cada uno de los puntos entre sí. $d_{i,j}=\mathcal{D}(X_i,X_j)$, $d_{i,j}=\mathcal{N}(X_i-X_j)$

- Agrupamiento jerárquico Divisivo: un solo grupo primero, luego en dos, luego en 3, hasta llegar tantos conjuntos como número de datos.
- Agrupamiento jerárquico Aglomerativo: lo mismo que el divisivo, pero al revés.

**Un buen agrupamiento**
- Minimizar distancias intragrupo (dentro del mismo grupo)
- Maximizar distancias intergrupo (entre grupos)

**Dendrogramas**
- Representación gráfica de cómo se observa un agrupamiento
- Visualización de las distancias entre los grupos, ya que se enlazan primero los grupos con menor distancia.

**Algoritmo de agrupamiento *k-means***

**Construcción de *k* grupos**
- Cada grupo tendrá un prototipo: 
$$
\mathbf{p}_\textbf{i}=\frac{\sum p_{i,j}x_j}{\sum p_{i,j}}
$$
- Cada elemento pertenecerá al grupo del prototipo más cercano:
$$
p_{i,j} =
\begin{cases}
1 & \text{si } \mathbf{p}_i \text{ es el prototipo más cercano a } \mathbf{x}_j \\
0 & \text{si otro caso}
\end{cases}
$$
- Minimizar:
$$
J = \sum_{i=1}^{k} \sum_{j=1}^{n} p_{i,j} \left\| \mathbf{p}_i - \mathbf{x}_j \right\|
$$





## Reglas de asociación

**Principios**
Tabla transaccional: Una tabla con banderas de 1 y 0.

Una regla de asociación tiene la forma: Antecedente -> Consecuente

Antecedente y Consecuente son conjuntos de ítems. Ej: $R:\left\{ x_1,x_3 \right\}\Rightarrow \left\{ x_2 \right\}$

Interpretación: En toda transacción que sucede el antecedente, también sucede el consecuente. (No necesariamente es una causalidad)

**Retícula de conjuntos de ítems *Item set Laticce***
![Item set Laticce]("/Módulo%204/ItemsetLattice.PNG")

La cantidad de las posibles reglas de asociación depende de el número de variables involucradas.

**¿Qué reglas son válidas?**
- Soporte de un itemset:
    - $sop(I)$ es el número de transacciones que contienen a $I$.
- Itemsets frecuentes:
    - itemsets cuyo soporte es mayor que un umbral $s_{min}$
- Soporte de la regla $R:A\to C$ (qué tantas veces sucede):
$$
sop(R)=\frac{sop(A\cup B)}{sop(\emptyset)}
$$
- Confianza de la regla $R:A\to C$ (probabilidad condicional de que suceda):
$$
conf(R)=\frac{sop(A\cup B)}{sop(A)}
$$

**Inducción de reglas de asociación**

Encontrar el conjunto de reglas de asociación cuyo soporte y confianza excedan unos umbrales específicos: $s_{min}, c_{min}$.

**Reglas de asociación para tablas no transaccionales**
- Definir intervalos para cada uno de los ítems.
- Crear una tabla transaccional para cada ítem para ver si pertenece al intervalo o no.
- Aplicar el algoritmo para encontrar las reglas de asociación.

**Un ejemplo de aplicación: Análisis de fallas en circuitos de distribución**
- Información disponible para una colección de fallas:
    - Fecha y hora
    - Duración
    - Circuito en que ocurrió
    - Causa de la falla
    - Tipo de falla
- Información disponible para cada circuito:
    - Ubicación
    - Potencia
    - Número de transformadores
    - Longitud
    - Número y tipo de usuarios

Al aplicar reglas de asociación, se concluyó:
- Circuitos cortos tienen fallas que duran poco tiempo
- Circuitos con pocos transformadores fallan pocas veces
