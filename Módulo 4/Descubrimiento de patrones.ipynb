**Algoritmos de arupamiento**
- Matriz de distancias: indica la distancia entre cada uno de los puntos entre sí. $d_{i,j}=\mathcal{D}(X_i,X_j)$, $d_{i,j}=\mathcal{N}(X_i-X_j)$

- Agrupamiento jerárquico Divisivo: un solo grupo primero, luego en dos, luego en 3, hasta llegar tantos conjuntos como número de datos.
- Agrupamiento jerárquico Aglomerativo: lo mismo que el divisivo, pero al revés.

**Un buen agrupamiento**
- Minimizar distancias intragrupo (dentro del mismo grupo)
- Maximizar distancias intergrupo (entre grupos)

**Dendrogramas**
- Representación gráfica de cómo se observa un agrupamiento
- Visualización de las distancias entre los grupos, ya que se enlazan primero los grupos con menor distancia.

**Algoritmo de agrupamiento *k-means***

**Construcción de *k* grupos**
- Cada grupo tendrá un prototipo: 
$$
\mathbf{p}_\textbf{i}=\frac{\sum p_{i,j}x_j}{\sum p_{i,j}}
$$
- Cada elemento pertenecerá al grupo del prototipo más cercano:
$$
p_{i,j} =
\begin{cases}
1 & \text{si } \mathbf{p}_i \text{ es el prototipo más cercano a } \mathbf{x}_j \\
0 & \text{si otro caso}
\end{cases}
$$
- Minimizar:
$$
J = \sum_{i=1}^{k} \sum_{j=1}^{n} p_{i,j} \left\| \mathbf{p}_i - \mathbf{x}_j \right\|
$$
