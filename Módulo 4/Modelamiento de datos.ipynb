# Modelamiento de datos

## El modelamiento aplicado a los datos

**Sistemas y modelos**
- Sistema: es "algo" bajo estudio
- Modelo: es una representación del sistema sobre la que podemos experimentar.

**Sobre los modelos**
- Tipos de modelos: físicos, mentales, gráficos, verbales, matemáticos, de software.
- Un mismo sistema puede representarse por muchos modelos
- ¿Cuál  es el mejor modelo?
    - Utilidad, que sirva, que alcance a representar
    - Parsimonia, que sea sencillo

**Modelamiento en la ciencia de datos**
Sistema -> Modelo $F(x)$ Implementación en software.

**Proceso genérico**
- Seleccionar la *clase de modelo* o *arquitectura*.
- Selecciona la *función de error*.
- Aplicar el *algoritmo*.
- *Validar* el resultado.
    - Generalización
    - Parsimonia
    - Interpretabilidad
    - Precisión

## Funciones de error

**El objetivo es minimizar el *error***

$f:\mathcal{M}\to \mathbb{R}$

$f$ es la *Función Objetivo*

$\mathcal{M} =$ Conjunto de todos los modelos de la clase seleccionada

**Error cuadrático medio**

$\varepsilon=\sqrt{\sum_{i=1}^{n}e_i{^2}}$

Dónde  $e_i=\tilde{y_i}-y_i$

El error es una distancia. Circunferencia unitaria para distintas Normas

**Errores de clasificación**

Matriz de costos, se le asigna un costo a cada equivocación.

$loss(C_i|E)=\sum_{i=1}^{m}P(C_j|E)c_{j,i}$

Matriz de confusión, aplicado a dos clases.

$Recall=\frac{VP}{VP + FN}$

$Especificidad=\frac{VN}{VN + FP}$

$Precisión=\frac{VP}{VP + FP}$

$Exactitud=\frac{VP + VN}{VP + FN + FP + VN}$

$F=2\frac{Precisión \times Recall}{Precisión + Recall}$

**Curva ROC (Receiver Operating Characteristic)**

Clasificador Parámetro $\sigma$.

Con la curva ROC, se busca que esté más alejado de los Falsos Positivos y esté más cerca a la proporción de los verdaderos positivos.

AUC: Área Bajo la Curva ROC

**Estategias de validación**

Separar los datos disponibles en dos grupos, uno para entrenar y otro para validar (testear).

Error de entrenamiento.

Error de prueba.

**Subentrenamiento vs Sobreentrenamiento**

Un modelo que generaliza bien, es un modelo cuyos errores de entrenamiento y prueba son comparables.
